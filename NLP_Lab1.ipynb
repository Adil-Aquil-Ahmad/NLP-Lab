```python
import nltk
import string
import random
import pandas as pd
import numpy as np

resources = ['brown', 'gutenberg', 'punkt', 'punkt_tab', 'stopwords',
             'wordnet', 'omw-1.4', 'words']
for res in resources:
    nltk.download(res, quiet=True)

random.seed(42)
print("Setup complete. All NLTK resources downloaded.")
```

```python
from nltk.corpus import brown, stopwords
from nltk.tokenize import sent_tokenize, word_tokenize

brown_text = ' '.join(brown.words(categories='news')[:2000])

sentences = sent_tokenize(brown_text)[:20]
print(f"Total sentences selected: {len(sentences)}")
print("\nExample sentence:\n", sentences[0])
```

```python
stop_words = set(stopwords.words('english'))
punctuations = set(string.punctuation)

def preprocess_sentence(sentence):
    """Tokenize, lowercase, and remove punctuation + stop words."""
    word_tokens = word_tokenize(sentence)
    lower_tokens = [w.lower() for w in word_tokens]
    cleaned_tokens = [w for w in lower_tokens
                       if w.isalpha() and w not in stop_words]
    return word_tokens, lower_tokens, cleaned_tokens
```

```python
results_q1 = []
for i, sent in enumerate(sentences[:5], start=1):
    word_tokens, lower_tokens, cleaned_tokens = preprocess_sentence(sent)
    results_q1.append({
        'Sentence #': i,
        'Original Sentence': sent,
        'Word Tokens': word_tokens,
        'Cleaned Tokens': cleaned_tokens
    })
    print(f"--- Sentence {i} ---")
    print("Original :", sent)
    print("Word Tokens:", word_tokens)
    print("Cleaned Tokens (lowercase, no punctuation/stopwords):", cleaned_tokens)
    print()

df_q1 = pd.DataFrame(results_q1)
df_q1
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

all_alpha_words = [w.lower() for w in brown.words(categories='news')
                    if w.isalpha() and w.lower() not in stop_words and len(w) > 2]

seen = set()
unique_words = []
for w in all_alpha_words:
    if w not in seen:
        seen.add(w)
        unique_words.append(w)

sample_words = unique_words[:30]

required_words = ['playing', 'studies', 'running', 'better', 'cars']
for w in required_words:
    if w not in sample_words:
        sample_words.append(w)

print(f"Total words selected: {len(sample_words)}")
print(sample_words)
```

```python
ps = PorterStemmer()
stemmed_words = [ps.stem(w) for w in sample_words]

lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(w, pos='v') for w in sample_words]

df_q2 = pd.DataFrame({
    'Original': sample_words,
    'Stemmed': stemmed_words,
    'Lemmatized': lemmatized_words
})
df_q2
```

```python
from nltk.corpus import words as words_corpus

vocabulary = set(w.lower() for w in words_corpus.words())
print(f"Vocabulary size: {len(vocabulary)}")

misspelled_words = ['recieve', 'definately', 'seperate', 'occured', 'adress',
                     'wich', 'beleive', 'goverment', 'tommorow', 'freind']
print("Misspelled words:", misspelled_words)
```

```python
def correct_word(word, vocab, max_candidates=200):
    """Return the closest vocabulary word using Levenshtein (edit) distance.
    Narrows the search space first (by first letter and similar length) for speed."""
    word = word.lower()
    if word in vocab:
        return word, 0

    candidates = [w for w in vocab
                  if w and w[0] == word[0] and abs(len(w) - len(word)) <= 2]
    if not candidates:
        candidates = list(vocab)

    best_word, best_dist = None, float('inf')
    for w in candidates:
        d = nltk.edit_distance(word, w)
        if d < best_dist:
            best_word, best_dist = w, d
    return best_word, best_dist

correction_results = []
for mw in misspelled_words:
    correction, dist = correct_word(mw, vocabulary)
    correction_results.append({'Misspelled': mw, 'Suggested Correction': correction, 'Edit Distance': dist})

df_q3 = pd.DataFrame(correction_results)
df_q3
```

```python
example_sentences = [
    "I would like to recieve the package tommorow.",
    "Please definately seperate the two documents.",
    "The goverment occured a major policy change.",
    "This is the correct adress for my freind.",
    "I beleive wich option is best for the project."
]

def correct_sentence(sentence):
    tokens = word_tokenize(sentence)
    corrected_tokens = []
    for t in tokens:
        if t.isalpha() and t.lower() not in vocabulary:
            corrected, _ = correct_word(t, vocabulary)
            corrected_tokens.append(corrected.capitalize() if t[0].isupper() else corrected)
        else:
            corrected_tokens.append(t)
    return ' '.join(corrected_tokens)

for sent in example_sentences:
    corrected = correct_sentence(sent)
    print("Original : ", sent)
    print("Corrected : ", corrected)
    print()
```

```python
from nltk.util import ngrams
from nltk import FreqDist

raw_tokens = brown.words(categories='news')[:10000]

clean_corpus_tokens = [w.lower() for w in raw_tokens if w.isalpha()]
print(f"Total tokens after cleaning: {len(clean_corpus_tokens)}")
print("Sample tokens:", clean_corpus_tokens[:20])
```

```python
unigrams = list(ngrams(clean_corpus_tokens, 1))
bigrams  = list(ngrams(clean_corpus_tokens, 2))
trigrams = list(ngrams(clean_corpus_tokens, 3))

print(f"Number of unigrams: {len(unigrams)}")
print(f"Number of bigrams : {len(bigrams)}")
print(f"Number of trigrams: {len(trigrams)}")
```

```python
unigram_freq = FreqDist(unigrams)
bigram_freq  = FreqDist(bigrams)
trigram_freq = FreqDist(trigrams)

top_unigrams = unigram_freq.most_common(10)
top_bigrams  = bigram_freq.most_common(10)
top_trigrams = trigram_freq.most_common(10)

print("Top 10 Unigrams:")
for gram, freq in top_unigrams:
    print(f"  {' '.join(gram):20s} -> {freq}")

print("\nTop 10 Bigrams:")
for gram, freq in top_bigrams:
    print(f"  {' '.join(gram):20s} -> {freq}")

print("\nTop 10 Trigrams:")
for gram, freq in top_trigrams:
    print(f"  {' '.join(gram):25s} -> {freq}")
```

```python
df_unigrams = pd.DataFrame([(' '.join(g), f) for g, f in top_unigrams], columns=['Unigram', 'Frequency'])
df_bigrams  = pd.DataFrame([(' '.join(g), f) for g, f in top_bigrams],  columns=['Bigram', 'Frequency'])
df_trigrams = pd.DataFrame([(' '.join(g), f) for g, f in top_trigrams], columns=['Trigram', 'Frequency'])

print("Unigram Frequency Table")
display(df_unigrams)
print("\nBigram Frequency Table")
display(df_bigrams)
print("\nTrigram Frequency Table")
display(df_trigrams)
```
